### 6.2 · Etapa B1: Análisis estático del costo hídrico y productivo de diluir $\text{ClO}_4^-$

## B1. ¿Cuánta agua haría falta para compensar un aumento de $\text{ClO}_4^-$? (Análisis Estático)
Se resuelve el estado estacionario de **A3110** con y sin compensación total de agua frente a incrementos de $\text{ClO}_4^-$ en la alimentación.


```python
rows = []
for f in (1.05, 1.10, 1.25, 1.50):
    R = f * (1 + R0) - 1
    Qd = R * qs0
    d = {**d0, "CP_stock": f * d0["CP_stock"]}
    xs, us, ds = pl.steady_state(alpha=0.25, d=d)
    ms = pl.metrics(xs, us, ds)
    xc, uc, dc = pl.steady_state(alpha=0.25, d=d, Q_dil=Qd)
    mc = pl.metrics(xc, uc, dc)
    rows.append({
        "ClO4 de entrada x": f,
        "dilución necesaria [m3/h]": 3600 * Qd,
        "veces la nominal": Qd / q0,
        "ppm sin comp.": ms["ppm_ClO4"],
        "Y sin comp. [%]": ms["recovery_pct"],
        "ppm con comp.": mc["ppm_ClO4"],
        "Y con comp. [%]": mc["recovery_pct"],
    })
req = pd.DataFrame(rows).set_index("ClO4 de entrada x")
req.round(2)
```

### 6.3 · Etapa B2: Cierre del lazo esclavo `FIRC 3103` frente a perturbación de válvula ($-20\%$)

## B2. Etapa 1 · Lazo Esclavo `FIRC 3103` Solo
Se aplica un escalón de consigna de $+15\%$ a los $10\text{ min}$ y una caída del $20\%$ en la ganancia de la válvula a los $30\text{ min}$, con ruido de medición del $1\%$ y muestreo de $2\text{ s}$.


```python
tun2, _ = C.autotune(pl, x0, u0, d0, dt_s=2.0)
flow_sp = lambda t: q0 * (1.15 if t >= 600 else 1.0)
dist1 = lambda t: {**d0, "valve_gain": 0.8 if t >= 1800 else 1.0}
cl = simulate(
    pl, x0, 3600, 2.0,
    C.PlantController(pl, u0, sp, tun2, enabled=("flow",), flow_sp=flow_sp, noise_rel={"flow": 0.01}, seed=1),
    disturbance=dist1, log_every=5
)
ol = simulate(
    pl, x0, 3600, 2.0,
    lambda t, x, pl_: {**u0, "Q_dil_cmd": flow_sp(t)},
    disturbance=dist1, log_every=5
)
late = lambda df: 100 * abs(df.Qd.values[df.t_h.values > 0.9].mean() / (1.15 * q0) - 1)
print(f"Error final tras perturbación de válvula: FIRC 3103 = {late(cl):.2f} % | Válvula manual = {late(ol):.2f} %")

fig, ax = plt.subplots(figsize=(7.5, 3.5))
ax.plot(cl.t_h * 60, 3600 * cl.Qd, label="Con FIRC 3103 (lazo cerrado)", color="tab:blue")
ax.plot(ol.t_h * 60, 3600 * ol.Qd, "--", label="Sin FIRC 3103 (válvula manual)", color="tab:red")
ax.plot(cl.t_h * 60, [3600 * flow_sp(t) for t in cl.t_h * 3600], "k:", label="Consigna F_303_sp")
ax.set(xlabel="Tiempo [min]", ylabel="Caudal dilución [m³/h]", title="Etapa 1: Desempeño del esclavo FIRC 3103")
ax.legend(fontsize=8)
fig.tight_layout()
fig.savefig("../docs/fig/06_etapa1_FIRC.png", dpi=150)
plt.show()
```

### 6.4 · Etapa B3: Cierre del controlador de razón `FFIC 3102` ante cambio de carga (`SSKd`)

## B3. Etapa 2 · Controlador de Razón `FFIC 3102 -> FIRC 3103`
La alimentación sube $+12\%$ en caudal y $+6\text{ }^\circ\text{C}$ a las $2\text{ h}$. Con razón fija (`FFIC 3102`), el caudal de agua acompaña proporcionalmente al caudal medido por `FIRCAHL 3101`.

### 6.5 · Etapa B4: Analizador `AIT 3105` y barrido de tiempo muerto ($\theta_{\text{AIT}} \in [0, 2\text{ h}]$)

## B4. Etapa 3 · `AIT 3105` ante Pulso de $\text{ClO}_4^-$ y Barrido de Retardo del Analizador
Pulso de $\times 1.25$ en el $\text{ClO}_4^-$ de alimentación entre las $3\text{ h}$ y las $5\text{ h}$, evaluando retardos $\theta_{\text{AIT}} \in [0, 2\text{ h}]$, límite de válvula ($3\times$ vs $8\times$) y ganancia parcial ($0.3$).


```python


def pulso(t):
    return d0["CP_stock"] * (1.25 if 3 * 3600 <= t < 5 * 3600 else 1.0)

dist3 = lambda t: {**d0, "CP_stock": pulso(t)}
H = 14 * 3600

def correr_ait(delay=None, cap=8.0, gain=1.0):
    if delay is None:
        ctrl = C.PlantController(pl, u0, sp, tun, enabled=("T", "V", "flow"))
    else:
        ctrl = C.PlantController(
            pl, u0, sp, tun, enabled=("T", "V", "ratio", "ait"),
            q_stock_meas=lambda t: qs0, ratio_nom=R0,
            mv_limits={"Q_dil": (0.0, cap * q0)},
            ait={"cp_true": pulso, "delay_s": delay, "gain": gain, "R_max": 3 * cap * R0}
        )
    return simulate(pl, x0, H, DT, ctrl, disturbance=dist3, log_every=12)

casos = {
    "sin AIT (dilución fija)": correr_ait(None),
    "AIT, retardo 0": correr_ait(0.0),
    "AIT, retardo 15 min": correr_ait(900.0),
    "AIT, retardo 1 h": correr_ait(3600.0),
    "AIT, retardo 2 h": correr_ait(7200.0),
    "AIT, retardo 0, válvula 3x": correr_ait(0.0, cap=3.0),
    "AIT, retardo 0, ganancia 0.3": correr_ait(0.0, gain=0.3),
}

fig, ax = plt.subplots(1, 3, figsize=(15, 4.2), sharex=True)
cols = ["tab:gray", "tab:blue", "tab:cyan", "tab:orange", "tab:red", "tab:purple", "tab:green"]
rows = []
for (name, df), col in zip(casos.items(), cols):
    ls = "--" if name.startswith("sin") else "-"
    ax[0].plot(df.t_h, df.ppm_ClO4, color=col, ls=ls, label=name)
    ax[1].plot(df.t_h, 3600 * df.Qd, color=col, ls=ls)
    ax[2].plot(df.t_h, df.recovery_pct, color=col, ls=ls)
    h = df.t_h.values
    ppm = df.ppm_ClO4.values
    Qd = 3600 * df.Qd.values
    rows.append({
        "caso": name,
        "ppm pico": ppm.max(),
        "horas sobre 40 ppm": float(np.trapezoid((ppm > 40).astype(float), h)),
        "Y mínima [%]": df.recovery_pct.min(),
        "Y final [%]": df.recovery_pct.iloc[-1],
        "dilución máx [m3/h]": Qd.max(),
        "agua extra [m3]": float(np.trapezoid(Qd - 3600 * q0, h)),
    })
ax[0].axhline(p["spec_ClO4_ppm_max"], color="r", lw=1)
ax[0].legend(fontsize=7)
for a in ax:
    a.axvspan(3, 5, color="orange", alpha=0.12)
ax[0].set(xlabel="t [h]", ylabel="ppm", title="ClO4 en producto (sombra = pulso)")
ax[1].set(xlabel="t [h]", ylabel="m³/h", title="Agua de dilución (FIRC 3103)")
ax[2].set(xlabel="t [h]", ylabel="%", title="Recuperación de KNO3")
fig.tight_layout()
fig.savefig("../docs/fig/05_AIT3105_retardo.png", dpi=150)
plt.show()

pd.DataFrame(rows).set_index("caso").round(2)
```

### 6.11 · Validación Multivariable (RGA / Niederlinski), *Lead-Lag*, *Override* y Restricción Dura de $\text{KClO}_4$ (EDT 4.2–4.3)

Consolida en un bloque reproducible los módulos avanzados incorporados en `msmpr_core/control.py` y `msmpr_core/dpbe.py`:
1. **Matriz RGA $3\times 3$ e Índice de Niederlinski ($NI$)** en `A3110` aislado vs. sub-bloques regulatorios $2\times 2$ (fundamento de la purga *plantwide* `AIC 3522 → FV 3522` en `B3520`).
2. **Compensador dinámico *Lead-Lag*** (`LeadLagCompensator`) y **selector *Override* de sobresaturación $\text{KNO}_3$** ($\sigma_{\min} = 0{,}02$).
3. **Restricción dura de saturación de $\text{KClO}_4$ calibrada** con el ensayo SQM ($\ln\Gamma_{\text{KClO}_4,0} = 1{,}571$, $T_{\text{sat}} = 41{,}29\text{ }^\circ\text{C}$) mediante `dilution_kclo4_hard_constraint()`.


```python
# ==============================================================================
# 6.11 · RGA / Niederlinski, Lead-Lag + Override y Restricción Dura de KClO4
# ==============================================================================
import numpy as np
import pandas as pd
try:
    from IPython.display import display
except ImportError:
    display = print
from msmpr_core import params, plant, dpbe, control as C

# 1. Matriz de ganancias G, RGA 3x3 e Índice de Niederlinski en A3110 (nivel V <-> Q_out cerrado)
pl_rga = plant.Plant(params.make_params())
x0_r, u0_r, d0_r = pl_rga.steady_state(alpha=0.25)
G_3, lam_3, cvs_3, mvs_3 = C.rga_from_plant(pl_rga, x0_r, u0_r, d0_r)
NI_3 = float(np.linalg.det(G_3) / np.prod(np.diag(G_3)))
y0_r = [pl_rga.metrics(np.asarray(x0_r, float), u0_r, d0_r).get(k, np.nan) if k not in ('T_C',) else x0_r[1] for k in cvs_3]
S_aut, sin_aut = C.mv_authority(G_3, [u0_r[m] for m in mvs_3], y0_r)
print(f'=== RGA 3x3 A3110 aislado (NI diagonal = {NI_3:.2f}) ===')
if sin_aut.any():
    print('AVISO: MV sin autoridad en el modelo aislado ->', [m for m, f in zip(mvs_3, sin_aut) if f],
          '| con V fijo el caudal de purga lo fija el balance de volumen, G es casi singular y RGA/NI NO son interpretables.')
    print('       La conclusión "purga plantwide" debe sustentarse con el modelo acoplado B3520 (alpha_purge actúa sobre el tanque), no con este NI.')
display(pd.DataFrame(lam_3, index=cvs_3, columns=mvs_3).round(3))

# --- Estructura corregida: el MV de calidad es el CAUDAL DE PURGA y el volumen se cierra con el agua de dilución ---
G_p, lam_p, cvs_p, mvs_p = C.rga_from_plant(pl_rga, x0_r, u0_r, d0_r, mv_names=("m_evap", "Q_out", "Q_purge"),
                                            volume_closure="Q_dil")
u0_p = dict(u0_r, Q_purge=u0_r["alpha"] * u0_r["Q_out"])
m0_p = pl_rga.metrics(np.asarray(x0_r, float), u0_p, d0_r)
y0_p = [x0_r[1] if k == "T_C" else m0_p[k] for k in cvs_p]
S_p, sin_p = C.mv_authority(G_p, [u0_p[m] for m in mvs_p], y0_p)
NI_p = float(np.linalg.det(G_p) / np.prod(np.diag(G_p)))
print(f"=== RGA 3x3 con Q_purge como MV de ppm y V cerrado con dilución (NI diagonal = {NI_p:.1f}) ===")
display(pd.DataFrame(lam_p, index=cvs_p, columns=mvs_p).round(3))
print("MV sin autoridad:", [m for m, f in zip(mvs_p, sin_p) if f])
print("Lectura: lambda(ppm, Q_purge) = %.3f -> el lazo de calidad (purga) es independiente de T y densidad: puede diseñarse como SISO. "
      "Q_out no tiene autoridad sobre la densidad en este modelo, así que el bloque T/densidad no es evaluable aquí." % lam_p[2, 2])


sub_rows = []
for nombre, ri, ci in [
    ('A. Térmico-Densidad (T_C, rho_pulpa) vs (m_evap, Q_dil)', [0, 1], [0, 1]),
    ('B. Térmico-Pureza (T_C, ppm_ClO4) vs (m_evap, Q_dil)', [0, 2], [0, 1]),
]:
    Gs = G_3[np.ix_(ri, ci)]
    ls = C.compute_rga(Gs)
    ni_s = float(np.linalg.det(Gs) / np.prod(np.diag(Gs)))
    sub_rows.append({'Sub-bloque 2x2 A3110': nombre, 'lambda_11': round(ls[0, 0], 3), 'lambda_12': round(ls[0, 1], 3), 'Niederlinski_NI': round(ni_s, 3), 'DIC_estable': ni_s > 0 and ls[0, 0] > 0})
print('\n=== Sub-bloques regulatorios 2x2 en A3110 ===')
display(pd.DataFrame(sub_rows))

# 2. Verificación de LeadLagCompensator y Override de sobresaturación KNO3
ll = C.LeadLagCompensator(tau_lead=60.0, tau_lag=20.0, gain=1.0, u0=0.0)
y_imp = ll.step(1.0, dt=1.0)
for _ in range(400):
    y_ss = ll.step(1.0, dt=1.0)
ov_demo = C.supersaturation_override_limit(C_in_kg_m3=630.0, C_star_kg_m3=600.0, R_cmd=0.08, sigma_min=0.02, R_min=0.0, R_max=0.15)
print(f'\nLead-Lag (tau_lead=60s, tau_lag=20s): impulso inicial={y_imp:.2f}, estacionario={y_ss:.3f}')
print(f'Override KNO3 (R_cmd=0.08): R_sigma_max={ov_demo["R_sigma_max"]:.4f}, R_applied={ov_demo["R_applied"]:.4f}, activo={ov_demo["override_active"]}')

# 3. Solubilidad calibrada de KClO4 (Apelblat + ensayo SQM) y restricción dura (EDT 4.3)
T_sat_sskd = dpbe.T_saturacion_KClO4(dpbe.COMP_SSKD_DEFAULT)
print(f'\n=== Restricción dura de KClO4 (T_sat SSKd base = {T_sat_sskd:.2f} °C, T_A3210 = 43.0 °C) ===')
comp_purged = dict(dpbe.COMP_SSKD_DEFAULT)
comp_purged['ClO4'] *= 0.75
casos_kclo4 = [
    ('A. Sostenible sin purga (techo 2,5 t/h)', dpbe.COMP_SSKD_DEFAULT, 2.50),
    ('B. Sostenible con purga AIC 3522 (-25% ClO4)', comp_purged, 2.50),
    ('C. Techo amplio sin purga (techo 40 t/h)', dpbe.COMP_SSKD_DEFAULT, 40.0),
]
filas_hc = []
for nombre, comp_c, f_max in casos_kclo4:
    o = C.dilution_kclo4_hard_constraint(
        C_in_kg_m3=660.0, C_star_kg_m3=600.0, R_cmd=0.50 / 378.18,
        T_stage2_C=43.0, comp_ref=comp_c, R_min=0.20 / 378.18, R_max=f_max / 378.18,
    )
    filas_hc.append({
        'Escenario (@43 °C)': nombre,
        'F_303_min_KClO4 [t/h]': round(o['F_303_kclo4_min_t_h'], 2),
        'F_303_aplicado [t/h]': round(o['F_303_applied_t_h'], 2),
        'T_sat_KClO4 [°C]': round(o['T_sat_KClO4_C'], 2),
        'Holgura [°C]': round(o['holgura_termica_C'], 2),
        'Requiere_purga_AIC3522': o['purge_required'],
        'Red_ClO4_requerida [%]': round(100.0 * o['clo4_reduction_needed_frac'], 1),
    })
display(pd.DataFrame(filas_hc))
```
